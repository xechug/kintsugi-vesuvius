```python
import sys
import os
import glob

print("Buscando archivos automáticamente en /kaggle/input...")
# Buscamos dinámicamente dónde ha montado Kaggle nuestro código
code_files = glob.glob('/kaggle/input/**/mamba_train.py', recursive=True)
if code_files:
    code_dir = os.path.dirname(code_files[0])
    print(f"✅ Código Kintsugi-Mamba encontrado en: {code_dir}")
    sys.path.insert(0, code_dir)
else:
    print("⚠️ CRÍTICO: No se encontró mamba_train.py. Asegúrate de añadir el dataset 'Vesuvius Mamba Code'.")

from mamba_train import train_model

# Buscamos dinámicamente los papiros de la competición
zarr_dirs = glob.glob('/kaggle/input/**/train/1/surface_volume', recursive=True)
if zarr_dirs:
    kaggle_zarr_dir = zarr_dirs[0]
    base_comp_dir = os.path.dirname(kaggle_zarr_dir)
    kaggle_mask_file = os.path.join(base_comp_dir, 'mask.png')
    kaggle_ink_file = os.path.join(base_comp_dir, 'inklabels.png')
    print(f"✅ Papiros oficiales encontrados en: {base_comp_dir}")
else:
    raise FileNotFoundError("No se encontró surface_volume. Asegúrate de añadir el dataset 'Vesuvius Challenge - Ink Detection'.")

guardar_en = "/kaggle/working/modelo_kintsugi_kaggle.pth"
print("\n🚀 Lanzando Simulador Kintsugi-Mamba en GPUs Nvidia T4...")

# Disparamos el entrenamiento
train_model(
    zarr_dir=kaggle_zarr_dir,
    mask_file=kaggle_mask_file,
    ink_file=kaggle_ink_file,
    save_path=guardar_en
)
```
